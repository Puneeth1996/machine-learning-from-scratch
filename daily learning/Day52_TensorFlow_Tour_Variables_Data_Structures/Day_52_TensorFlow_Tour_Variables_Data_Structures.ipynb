## 1. Import TensorFlow and NumPy

TensorFlow is primarily a numerical-computing framework designed for machine learning. Its basic numerical object is a **tensor**.

A tensor is similar to a NumPy array, but TensorFlow tensors can be efficiently processed on CPUs, GPUs, and other accelerators and can participate in automatic differentiation.


````python
import tensorflow as tf
import numpy as np

print("TensorFlow version:", tf.__version__)
print("NumPy version:", np.__version__)
````

## 2. What is a Tensor?

A tensor is a multidimensional array.

Examples:

- **Scalar:** rank 0 → `42`
- **Vector:** rank 1 → `[1, 2, 3]`
- **Matrix:** rank 2 → `[[1, 2], [3, 4]]`
- **Higher-dimensional tensor:** rank 3 or more

The three properties you should always remember are:

```text
rank  → number of dimensions
shape → size along each dimension
dtype → data type of elements
```


````python
scalar = tf.constant(42)
vector = tf.constant([1, 2, 3])
matrix = tf.constant([[1, 2], [3, 4]])
tensor_3d = tf.constant([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

for name, x in [
    ("scalar", scalar),
    ("vector", vector),
    ("matrix", matrix),
    ("tensor_3d", tensor_3d)
]:
    print(f"{name}:")
    print(" value:", x)
    print(" rank :", tf.rank(x).numpy())
    print(" shape:", x.shape)
    print(" dtype:", x.dtype)
    print()
````

## 3. Creating Tensors

TensorFlow provides many functions that feel familiar if you know NumPy.

Common functions:

- `tf.zeros()`
- `tf.ones()`
- `tf.fill()`
- `tf.range()`
- `tf.eye()`
- `tf.random.normal()`
- `tf.random.uniform()`


````python
zeros = tf.zeros((2, 3))
ones = tf.ones((2, 3))
filled = tf.fill((2, 3), 7)
sequence = tf.range(0, 10, delta=2)
identity = tf.eye(3)

print("Zeros:\n", zeros.numpy())
print("\nOnes:\n", ones.numpy())
print("\nFilled:\n", filled.numpy())
print("\nRange:", sequence.numpy())
print("\nIdentity:\n", identity.numpy())
````

## 4. Tensor Shape and Reshaping

A tensor's shape tells us how many elements exist along each dimension.

For example:

```text
shape = (2, 3, 4)
```

means:

- 2 blocks
- 3 rows per block
- 4 values per row

`tf.reshape()` changes the organization of the tensor without changing the number of elements.


````python
x = tf.range(24)

print("Original:")
print(x)
print("Shape:", x.shape)

x_2d = tf.reshape(x, (6, 4))
x_3d = tf.reshape(x, (2, 3, 4))

print("\n6 x 4:")
print(x_2d.numpy())

print("\n2 x 3 x 4:")
print(x_3d.numpy())
````

## 5. Indexing and Slicing

TensorFlow tensors support NumPy-like indexing and slicing.

Examples:

```python
x[0]
x[:, 1]
x[1:3]
x[..., -1]
```


````python
x = tf.constant([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("x:\n", x.numpy())

print("\nFirst row:", x[0].numpy())
print("Second column:", x[:, 1].numpy())
print("Rows 1 onward:\n", x[1:].numpy())
print("Last column:", x[..., -1].numpy())
````

## 6. TensorFlow Operations

TensorFlow provides operations for arithmetic, linear algebra, reductions, comparisons, and more.

Common examples:

- `tf.add`
- `tf.subtract`
- `tf.multiply`
- `tf.divide`
- `tf.square`
- `tf.sqrt`
- `tf.maximum`
- `tf.reduce_sum`
- `tf.reduce_mean`
- `tf.matmul`


````python
a = tf.constant([[1., 2.], [3., 4.]])
b = tf.constant([[5., 6.], [7., 8.]])

print("Addition:\n", tf.add(a, b).numpy())
print("\nElement-wise multiplication:\n", tf.multiply(a, b).numpy())
print("\nSquare of a:\n", tf.square(a).numpy())
print("\nMean of a:", tf.reduce_mean(a).numpy())
print("Sum of a:", tf.reduce_sum(a).numpy())

print("\nMatrix multiplication:")
print(tf.matmul(a, b).numpy())
````

## 7. TensorFlow Operators Look Like NumPy

TensorFlow also overloads Python operators, so you can often write:

```python
a + b
a * b
a / b
a ** 2
```

instead of explicitly calling `tf.add()`, `tf.multiply()`, etc.

This makes TensorFlow code feel very similar to NumPy.


````python
a = tf.constant([[1., 2.], [3., 4.]])
b = tf.constant([[10., 20.], [30., 40.]])

print("a + b:\n", (a + b).numpy())
print("\na * b:\n", (a * b).numpy())
print("\na / 2:\n", (a / 2).numpy())
print("\na ** 2:\n", (a ** 2).numpy())
````

## 8. Broadcasting

TensorFlow follows NumPy-style broadcasting rules.

For example, a matrix with shape `(2, 3)` can be added to a vector with shape `(3,)`.

The vector is conceptually broadcast across the rows.


````python
matrix = tf.constant([
    [1., 2., 3.],
    [4., 5., 6.]
])

bias = tf.constant([10., 20., 30.])

result = matrix + bias

print("Matrix shape:", matrix.shape)
print("Bias shape:", bias.shape)
print("\nResult:")
print(result.numpy())
````

## 9. TensorFlow and NumPy

TensorFlow and NumPy work together very well.

### NumPy → TensorFlow

```python
tf.constant(numpy_array)
```

### TensorFlow → NumPy

```python
tensor.numpy()
```

One important distinction is that TensorFlow tensors are designed to participate in TensorFlow's computation and automatic differentiation system, while NumPy arrays are NumPy objects.


````python
numpy_array = np.array([[1, 2], [3, 4]], dtype=np.float32)

tensor = tf.constant(numpy_array)
back_to_numpy = tensor.numpy()

print("NumPy array:")
print(numpy_array)
print("Type:", type(numpy_array))

print("\nTensorFlow tensor:")
print(tensor)
print("Type:", type(tensor))

print("\nBack to NumPy:")
print(back_to_numpy)
print("Type:", type(back_to_numpy))
````

## 10. Type Conversions

TensorFlow is more strict about data types than Python can sometimes appear to be.

For example, mixing `float32` and `float64` tensors may cause errors in operations where matching dtypes are required.

Use `tf.cast()` when you intentionally need to convert a tensor's dtype.


````python
x = tf.constant([1, 2, 3], dtype=tf.int32)

x_float = tf.cast(x, tf.float32)
x_float64 = tf.cast(x, tf.float64)

print("Original:", x, x.dtype)
print("float32:", x_float, x_float.dtype)
print("float64:", x_float64, x_float64.dtype)
````

## 11. TensorFlow Variables

This is one of the most important topics of this day.

A `tf.Tensor` is generally treated as an immutable value.

A `tf.Variable` is a mutable state container.

Neural networks need mutable state because model parameters such as **weights and biases must be updated during training**.

Typical pattern:

```python
weight = tf.Variable(initial_value)
weight.assign(new_value)
```

During training, an optimizer repeatedly updates variables to reduce the loss.


````python
weight = tf.Variable([1.0, 2.0, 3.0])

print("Initial weight:", weight.numpy())

weight.assign([10.0, 20.0, 30.0])

print("After assign:", weight.numpy())
````

## 12. `assign_add()` and `assign_sub()`

Variables support convenient in-place-style updates:

```python
variable.assign_add(...)
variable.assign_sub(...)
```

These are useful for counters and model parameters.


````python
counter = tf.Variable(0)

print("Initial:", counter.numpy())

counter.assign_add(5)
print("After +5:", counter.numpy())

counter.assign_sub(2)
print("After -2:", counter.numpy())
````

## 13. Why Variables Matter in Neural Networks

A simplified neural-network training step looks conceptually like this:

```text
Input X
   ↓
Weights W + Bias b
   ↓
Prediction
   ↓
Loss
   ↓
Gradients
   ↓
Optimizer
   ↓
Update W and b
```

`W` and `b` are TensorFlow variables.

For example:

```python
W = tf.Variable(...)
b = tf.Variable(...)
```

The optimizer modifies these variables repeatedly during training.


````python
# A tiny manually-updated linear model

W = tf.Variable(0.0)
b = tf.Variable(0.0)

x = tf.constant(2.0)
y = tf.constant(10.0)

learning_rate = 0.1

prediction = W * x + b
loss = (prediction - y) ** 2

print("Initial W:", W.numpy())
print("Initial b:", b.numpy())
print("Prediction:", prediction.numpy())
print("Loss:", loss.numpy())
````

## 14. TensorFlow Strings

TensorFlow also has a string tensor type.

Strings are not ordinary numerical tensors, but TensorFlow provides operations for processing them, especially in data pipelines.


````python
text = tf.constant(["TensorFlow", "Machine Learning", "Python"])

print(text)
print("dtype:", text.dtype)

lower = tf.strings.lower(text)
print("\nLowercase:")
print(lower.numpy())

lengths = tf.strings.length(text)
print("\nString lengths:")
print(lengths.numpy())
````

## 15. Ragged Tensors

A normal tensor requires a rectangular shape.

But real-world data may have different numbers of elements per row.

Example:

```text
[10, 20]
[30, 40, 50]
[60]
```

This cannot naturally be represented by a regular rectangular tensor without padding.

A **ragged tensor** allows rows to have different lengths.

This is useful for variable-length sequences such as:

- sentences with different numbers of words
- user histories
- sequences of events


````python
ragged = tf.ragged.constant([
    [10, 20],
    [30, 40, 50],
    [60]
])

print(ragged)
print("\nRow 0:", ragged[0].numpy())
print("Row 1:", ragged[1].numpy())
print("Row 2:", ragged[2].numpy())

print("\nRow lengths:", ragged.row_lengths().numpy())
````

## 16. Sparse Tensors

A sparse tensor is useful when most values are zero.

Instead of storing every zero, TensorFlow can store only the non-zero values and their locations.

Example:

```text
0 0 0 5
0 0 0 0
0 7 0 0
```

This can save memory when tensors are extremely sparse.

Sparse tensors are common in areas such as:

- recommendation systems
- text processing
- large categorical features
- graph-like data


````python
sparse = tf.sparse.SparseTensor(
    indices=[[0, 3], [2, 1]],
    values=[5, 7],
    dense_shape=[3, 4]
)

print("Sparse tensor:")
print(sparse)

print("\nDense representation:")
print(tf.sparse.to_dense(sparse).numpy())
````

## 17. TensorArray

`tf.TensorArray` is a TensorFlow data structure designed for storing tensors dynamically, especially inside TensorFlow control-flow operations.

You will encounter it less frequently in everyday Keras code, but it is useful to recognize the concept.

For many standard ML workflows, ordinary tensors, variables, ragged tensors, and sparse tensors are more important.


````python
ta = tf.TensorArray(dtype=tf.float32, size=3)

ta = ta.write(0, 10.0)
ta = ta.write(1, 20.0)
ta = ta.write(2, 30.0)

print("TensorArray values:")
print(ta.stack().numpy())
````

## 18. TensorFlow Sets

TensorFlow also has support for set-like operations.

These can be useful when working with unique categorical values or membership operations.


````python
a = tf.constant([1, 2, 3, 3])
b = tf.constant([3, 4, 5])

unique_a = tf.unique(a).y

print("Unique values in a:", unique_a.numpy())

set_result = tf.sets.union(
    tf.reshape(a, (1, -1)),
    tf.reshape(b, (1, -1))
)

print("\nUnion:")
print(tf.sparse.to_dense(set_result).numpy())
````

## 19. TensorFlow vs NumPy — Quick Comparison

| Concept | NumPy | TensorFlow |
|---|---|---|
| Main array type | `ndarray` | `Tensor` |
| Mutable parameter container | — | `tf.Variable` |
| GPU/accelerator support | Limited/general ecosystem | Built into TensorFlow |
| Automatic differentiation | No | Yes |
| ML model training | External libraries | Core capability |
| Ragged data | Usually custom/object approaches | `tf.RaggedTensor` |
| Sparse data | Sparse libraries | `tf.sparse` |

The important point is not that TensorFlow replaces NumPy.

Instead:

```text
NumPy       → general numerical computing
TensorFlow  → numerical computing + differentiation + ML training
```
