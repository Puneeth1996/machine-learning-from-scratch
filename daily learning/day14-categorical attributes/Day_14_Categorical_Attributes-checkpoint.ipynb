# Day 14 - Categorical Attributes
Learn how to turn text categories into numbers for machine learning.
In this notebook, we use a small dataset with `Color`, `Size`, and `Price`.

We will practice:
- Ordinal encoding for ordered categories like sizes
- One-hot encoding for unordered categories like colors
- Comparing dataset shape before and after encoding

## Step 1: Create a simple dataset
This dataset has two categorical columns:
- `Color`: no natural order (Red, Blue, Green)
- `Size`: ordered categories (Small, Medium, Large)

The `Price` column is numeric and does not need encoding.

```python
import pandas as pd

# Build a small example dataset
# Each row is one item with a color, size, and price.
df = pd.DataFrame({
    'Color': ['Red', 'Blue', 'Green', 'Blue'],
    'Size': ['Small', 'Medium', 'Large', 'Medium'],
    'Price': [100, 120, 150, 130]
})

df
```

## Step 2: Ordinal Encoding
Ordinal encoding maps categories to integers when there is a clear order.

For `Size`, we define:
- `Small` → 0
- `Medium` → 1
- `Large` → 2

This keeps the order information, so `Large` is still greater than `Small`.

```python
from sklearn.preprocessing import OrdinalEncoder

# Use an ordered category list to preserve the correct ranking.
enc = OrdinalEncoder(categories=[['Small', 'Medium', 'Large']])

# fit_transform returns a 2D array, so we pass df[['Size']] as a DataFrame.
df['Size_Ordinal'] = enc.fit_transform(df[['Size']])

# Show the dataframe with the new encoded size column.
df
```

## Step 3: One-Hot Encoding
One-hot encoding is used for unordered categories like `Color`.
It creates one binary column per category and avoids implying any order.

For `Color`, the result will be:
- `Color_Blue`
- `Color_Green`
- `Color_Red`

```python
from sklearn.preprocessing import OneHotEncoder

# OneHotEncoder creates binary columns for each unique category.
ohe = OneHotEncoder(sparse_output=False)
encoded = ohe.fit_transform(df[['Color']])

# Convert the result back into a DataFrame with readable column names.
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(['Color']))
encoded_df
```

## Step 4: Combine encoded columns and compare shapes
Now we add the new one-hot columns to the dataframe and remove the original `Color` column.
This shows how the dataset changes shape after encoding.

```python
# Combine the encoded color columns with the original dataframe.
final_df = pd.concat([df.drop(columns=['Color']), encoded_df], axis=1)

print('Original shape:', df.shape)
print('Encoded shape:', final_df.shape)

final_df
```

## What is going on in each cell?
- The first markdown cell explains the goal of the notebook.
- The dataset cell builds a small table and shows the raw categorical data.
- The ordinal encoding section explains why `Size` is ordered and shows the transformed values.
- The one-hot encoding section explains why `Color` is unordered and creates binary indicator columns.
- The final cell combines the data and prints the shape difference to compare before/after encoding.

## Practice ideas
- Try encoding `Size` with one-hot encoding instead of ordinal encoding.
- Add a new color like `Yellow` and rerun the one-hot encoder.
- Add a new size category like `X-Large`, then update the ordinal category list.
- Compare `df.shape` before and after encoding to see how many columns were added.

### Summary
- **Ordinal Encoding**: use for categories with order.
- **One-Hot Encoding**: use for categories without order.
- After encoding, the number of columns often increases, especially with one-hot encoding.
- One-hot avoids false numeric relationships between categories.
