## Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
)
from sklearn.multiclass import OneVsRestClassifier
from sklearn.multioutput import MultiOutputClassifier
from sklearn.neighbors import KNeighborsClassifier

np.random.seed(42)
```

# 1. Load the Digits Dataset

The digits dataset contains **8×8 grayscale images** of handwritten digits (0–9).


```python
digits = load_digits()

X = digits.data
y = digits.target

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
print("Classes:", np.unique(y))
```

```python
plt.figure(figsize=(3, 3))
plt.imshow(digits.images[0], cmap="gray")
plt.title(f"Label: {y[0]}")
plt.axis("off")
plt.show()
```

### Key Takeaways
- Each image is represented by **64 pixel features**.
- The task is **multiclass classification** because there are 10 possible classes.


# 2. Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Train size:", X_train.shape[0])
print("Test size :", X_test.shape[0])
```

# 3. Train a Classifier

We use **SGDClassifier**, which is the same family of classifier discussed in the book.


```python
sgd_clf = SGDClassifier(random_state=42)
sgd_clf.fit(X_train, y_train)

y_pred = sgd_clf.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
```

### Key Takeaways
- Accuracy gives a quick overall score.
- Accuracy alone may hide **which classes are failing**.
- We need deeper analysis.


# 4. Confusion Matrix (Error Analysis)

A confusion matrix shows **true labels vs predicted labels**.


```python
cm = confusion_matrix(y_test, y_pred)

fig, ax = plt.subplots(figsize=(7, 7))
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(ax=ax, cmap="Blues", colorbar=False)
plt.title("Confusion Matrix")
plt.show()
```

### Interpretation

- Diagonal cells = correct predictions.
- Off-diagonal cells = mistakes.
- Larger off-diagonal values indicate confusing digit pairs.


# 5. Precision, Recall, and F1-score

```python
print(classification_report(y_test, y_pred))
```

```python
macro_precision = precision_score(y_test, y_pred, average="macro")
macro_recall = recall_score(y_test, y_pred, average="macro")
macro_f1 = f1_score(y_test, y_pred, average="macro")

print("Macro Precision:", round(macro_precision, 4))
print("Macro Recall   :", round(macro_recall, 4))
print("Macro F1-score :", round(macro_f1, 4))
```

### What these metrics mean

- **Precision**: Of predicted positives, how many were correct?
- **Recall**: Of actual positives, how many were found?
- **F1-score**: Harmonic mean of precision and recall.

The book emphasizes that these metrics are often more informative than accuracy.


# 6. Normalize the Confusion Matrix

Normalization highlights the **error rate per class**.


```python
cm_normalized = cm.astype("float") / cm.sum(axis=1, keepdims=True)

fig, ax = plt.subplots(figsize=(7, 7))
im = ax.imshow(cm_normalized, cmap="Blues")
ax.set_title("Normalized Confusion Matrix")
ax.set_xlabel("Predicted label")
ax.set_ylabel("True label")
ax.set_xticks(range(10))
ax.set_yticks(range(10))
plt.colorbar(im)
plt.show()
```

### Key Takeaways
- Rows sum to 1.
- Dark off-diagonal cells reveal classes with higher confusion.
- This is one of the core **error-analysis techniques** from Chapter 3.


# 7. Inspect Misclassified Images

Let's visualize a few mistakes.


```python
mis_idx = np.where(y_test != y_pred)[0][:9]

fig, axes = plt.subplots(3, 3, figsize=(6, 6))
for ax, idx in zip(axes.ravel(), mis_idx):
    ax.imshow(X_test[idx].reshape(8, 8), cmap="gray")
    ax.set_title(f"T:{y_test[idx]} P:{y_pred[idx]}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

### Why inspect mistakes?

The book recommends **looking at misclassified examples** to understand:
- ambiguous handwriting,
- poor image quality,
- systematic confusion between classes.


# 8. Multilabel Classification

A sample can belong to **multiple labels simultaneously**.

We create two labels:
- **large_digit** → digit ≥ 7
- **odd_digit** → digit is odd


```python
y_large = (y >= 7)
y_odd = (y % 2 == 1)

Y_multilabel = np.c_[y_large, y_odd]

print("First 5 multilabel targets:")
print(Y_multilabel[:5])
```

```python
X_train_ml, X_test_ml, Y_train_ml, Y_test_ml = train_test_split(
    X, Y_multilabel, test_size=0.2, random_state=42
)

ovr_clf = OneVsRestClassifier(SGDClassifier(random_state=42))
ovr_clf.fit(X_train_ml, Y_train_ml)

Y_pred_ml = ovr_clf.predict(X_test_ml)

print("Subset accuracy:", round(accuracy_score(Y_test_ml, Y_pred_ml), 4))
```

```python
results = pd.DataFrame({
    "True large": Y_test_ml[:, 0][:10],
    "Pred large": Y_pred_ml[:, 0][:10],
    "True odd": Y_test_ml[:, 1][:10],
    "Pred odd": Y_pred_ml[:, 1][:10],
})

results
```

### Key Takeaways
- Each image now has **two target values**.
- Multilabel classification predicts multiple labels for the same sample.
- Example: digit **9** → `[True, True]`.


# 9. Multioutput Classification

Multioutput classification predicts **multiple outputs** for each sample.

We create two outputs:
1. original digit
2. parity (0 = even, 1 = odd)


```python
y_digit = y
y_parity = y % 2

Y_multioutput = np.c_[y_digit, y_parity]

print("Example targets:")
print(Y_multioutput[:5])
```

```python
X_train_mo, X_test_mo, Y_train_mo, Y_test_mo = train_test_split(
    X, Y_multioutput, test_size=0.2, random_state=42
)

multi_clf = MultiOutputClassifier(KNeighborsClassifier())
multi_clf.fit(X_train_mo, Y_train_mo)

Y_pred_mo = multi_clf.predict(X_test_mo)

digit_acc = accuracy_score(Y_test_mo[:, 0], Y_pred_mo[:, 0])
parity_acc = accuracy_score(Y_test_mo[:, 1], Y_pred_mo[:, 1])

print("Digit accuracy :", round(digit_acc, 4))
print("Parity accuracy:", round(parity_acc, 4))
```

```python
comparison = pd.DataFrame({
    "True Digit": Y_test_mo[:10, 0],
    "Pred Digit": Y_pred_mo[:10, 0],
    "True Parity": Y_test_mo[:10, 1],
    "Pred Parity": Y_pred_mo[:10, 1],
})

comparison
```

### Multilabel vs Multioutput

| Type | Output |
|---|---|
| Multilabel | Multiple **binary labels** |
| Multioutput | Multiple targets, each may have several classes |

The book treats multioutput classification as a more general setting.
